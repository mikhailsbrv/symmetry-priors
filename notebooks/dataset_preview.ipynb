```python
import torch
import matplotlib.pyplot as plt

dataset = torch.load('../data/letters.pt') 
```

```python
dataset.keys()
```

```python
dataset['images'][0].shape
```

```python
aA = dataset['images'][0].squeeze(1)
```

```python
a, A = aA[0], aA[1]
```

```python
image_a, image_A = a.numpy(), A.numpy()

fig, axes = plt.subplots(1, 2)

axes[0].imshow(image_a, cmap='gray')
axes[0].axis('off')

axes[1].imshow(image_A, cmap='gray')
axes[1].axis('off')
```

```python
max(max(c) for c in image_a)
```

```python
min(min(c) for c in image_a)
```

```python
images = dataset['images'].squeeze(2)
```

```python
images.shape
```

```python
fig, axes = plt.subplots(images.shape[0], 2, figsize=(6, 3 * images.shape[0]))

for i, letter in enumerate(images):     
    axes[i, 0].imshow(letter[0].numpy(), cmap='gray')
    axes[i, 0].axis('off')
    
    axes[i, 1].imshow(letter[1].numpy(), cmap='gray')
    axes[i, 1].axis('off')

plt.tight_layout()
plt.show()
    
```
